```python
#Two Small Functions
def label_amount(amount):
    """Label a transaction amount in KES (thresholds are our own, for illustration)."""
    if amount < 500:
        return "Low"
    elif amount < 5000:
        return "Medium"
    else:
        return "High"

def meets_size(df, min_rows=15000, min_cols=10):
    """Return True if the DataFrame has enough rows and columns."""
    rows, cols = df.shape
    return rows >= min_rows and cols >= min_cols

for sample in [120, 1500, 25000]:
    print(sample, "->", label_amount(sample))
```
